# Table 6: NSL-KDD Per-Attack Classification Report
### *Independent Replication of Sharma et al. (2024)*
**Paper Reference**: *Expert Systems with Applications* 238 (2024) 121751, Table 6

Evaluates **DoS**, **Normal**, **Probe**, **R2L**, and **U2R** attacks across **DNN**, **1D-CNN**, and **2D-CNN**.

```python
import os, sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path(os.getcwd()).resolve()
if (ROOT / 'results').is_dir():
    table6_csv = ROOT / 'results' / 'comparison' / 'table6_nsl_kdd_per_class.csv'
else:
    table6_csv = ROOT.parent / 'results' / 'comparison' / 'table6_nsl_kdd_per_class.csv'

df = pd.read_csv(table6_csv)
print("="*85)
print(" TABLE 6: NSL-KDD PER-ATTACK METRICS (PAPER VS OUR REPLICATION)")
print("="*85)
display(df)
```

```python
# Format side-by-side matching Sharma et al. Table 6 exactly
models = ['DNN', '1D-CNN', '2D-CNN']
attacks = ['DoS', 'Normal', 'Probe', 'R2L', 'U2R', 'Accuracy']

paper_cols = {}
repl_cols = {}
for m in models:
    for met in ['Precision', 'Recall', 'F1-Score']:
        paper_cols[(m, met)] = []
        repl_cols[(m, met)] = []

for a in attacks:
    for m in models:
        row = df[(df['Attack'] == a) & (df['Model'] == m)]
        paper_cols[(m, 'Precision')].append(f"{row['Paper_Precision'].values[0]:.2f}")
        paper_cols[(m, 'Recall')].append(f"{row['Paper_Recall'].values[0]:.2f}")
        paper_cols[(m, 'F1-Score')].append(f"{row['Paper_F1'].values[0]:.2f}")
        repl_cols[(m, 'Precision')].append(f"{row['Ours_Precision'].values[0]:.2f}")
        repl_cols[(m, 'Recall')].append(f"{row['Ours_Recall'].values[0]:.2f}")
        repl_cols[(m, 'F1-Score')].append(f"{row['Ours_F1'].values[0]:.2f}")

print("\n" + "="*90)
print(" PUBLISHED TABLE 6 (SHARMA ET AL. 2024)")
print("="*90)
pdf = pd.DataFrame(paper_cols, index=attacks)
pdf.columns = pd.MultiIndex.from_tuples(pdf.columns, names=['Model', 'Metric'])
display(pdf)

print("\n" + "="*90)
print(" REPLICATED TABLE 6 (OUR INDEPENDENT RUNS)")
print("="*90)
rdf = pd.DataFrame(repl_cols, index=attacks)
rdf.columns = pd.MultiIndex.from_tuples(rdf.columns, names=['Model', 'Metric'])
display(rdf)
```

```python
# Visualization Bar Plot
plot_df = df[df['Attack'] != 'Accuracy']
fig, axes = plt.subplots(1, 3, figsize=(17, 5), sharey=True)

for idx, m in enumerate(models):
    sub = plot_df[plot_df['Model'] == m]
    x = np.arange(len(sub))
    axes[idx].bar(x - 0.18, sub['Paper_F1'], 0.35, label='Paper', color='#2b5c8f')
    axes[idx].bar(x + 0.18, sub['Ours_F1'], 0.35, label='Ours', color='#28a745')
    axes[idx].set_title(f"{m} F1 Comparison", fontweight='bold')
    axes[idx].set_xticks(x)
    axes[idx].set_xticklabels(sub['Attack'])
    axes[idx].legend()
    axes[idx].grid(axis='y', linestyle='--', alpha=0.6)

plt.suptitle("NSL-KDD (Table 6) F1-Score: Published vs Ours", fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```
